###  To fix this issue You will fill these values with ........... for numeric and ......for categorical

```python
# trying to fill missing values in my dataset using for loop
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype in ['int64', 'float64']:
            # Fill numeric columns with the mean (or median)
            df[col].fillna(df[col].mean(), inplace=True)
        else:
            # Fill categorical columns with the mode (most frequent value)
            df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
#check the Nulls 
df.isnull().sum()
```

# <a id="14"> 14 )  .  What is the  Correlation between `specific numeric variables` in the dataset?</a>


```python
# Make visualization to easy get insight
df_clean = df[(df['member_age'] <= 80) & (df['duration_min'] <= 60)]

plt.figure(figsize=(10, 5))
sns.regplot(data=df_clean.sample(5000, random_state=42), x='member_age', y='duration_min',
            scatter_kws={'alpha':0.2}, line_kws={'color':'red'})

plt.title('Trip Duration (Min) vs Member Age', fontsize=14, fontweight='bold')
plt.xlabel('Member Age (Years)', fontsize=12)
plt.ylabel('Trip Duration (Minutes)', fontsize=12)
plt.show()

```

> ### <mark>Your analsys and exploration here </mark>
* There is **no Correlation** for any variable with time (`duration sec /minutes`).
* **Flat Trendline:** The red regression line remains completely horizontal across all ages, showing no linear relationship between age and trip duration.
* **Consistent Usage Pattern:** Most users across all age groups (from 20 to 70 years) complete their trips within **10 to 15 minutes**.
