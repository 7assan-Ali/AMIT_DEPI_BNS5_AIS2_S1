```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import mean_squared_error,r2_score
```

```python
x=np.array([[1],[2],[3],[4]])
y=np.array([2,4,6,8])
```

```python
#=========================
#linear Regression
#=========================
lin_reg = LinearRegression()
lin_reg.fit(x,y)
y_pred_lin = lin_reg.predict(x)
#=========================
#Ridge Regression with small alpha
ridge = Ridge(alpha = 1.0)
ridge.fit(x,y)
y_pred_ridge = ridge.predict(x)
#=========================
#Lasso Regression with large alpha
lasso = Lasso(alpha = 10)
lasso.fit(x,y)
y_pred_lasso = lasso.predict(x)
#=========================
print("=========================")
print("Linear Regression:")
print("MSE: ", mean_squared_error(y,y_pred_lin))
print("b: ", lin_reg.intercept_)
print("w: ", lin_reg.coef_)
#=========================
print("=========================")
print("Ridge Regression:")
print("MSE: ", mean_squared_error(y,y_pred_ridge))
print("b: ", ridge.intercept_)
print("w: ", ridge.coef_)
#=========================
print("=========================")
print("Lasso Regression:")
print("MSE: ", mean_squared_error(y,y_pred_lasso))
print("b: ", lasso.intercept_)
print("w: ", lasso.coef_)

```

```python
plt.figure(figsize=(10,8))
plt.subplot(1,3,2)
plt.scatter(x,y,color="blue",label="actual data")
plt.plot(x,y_pred_ridge,color="green",label="Ridge Regression")
plt.title("Linear Regression")
plt.legend()
```

```python
plt.scatter(x, y, color="blue", label="Actual data")
plt.plot(x, y_pred_lin, color="green", label="Linear Regression")
plt.plot(x, y_pred_ridge, color="red", label="Ridge Regression")
plt.plot(x, y_pred_lasso, color="orange", label="Lasso Regression")
plt.title("Comparison of linear ,Ridge and Lasso Regression")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
