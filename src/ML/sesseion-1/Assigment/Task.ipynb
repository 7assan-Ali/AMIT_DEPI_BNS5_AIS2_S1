```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
class LinearRegression:

    def __init__(self, learning_rate=0.01, n_iters=100):
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.weight = 0
        self.bias = 0
        self.sse_history = []

    def fit(self, x, y):
        x = np.array(x)
        y = np.array(y)
        n = len(x)

        for i in range(self.n_iters):
            y_pred = self.weight * x + self.bias
            error = y_pred - y

            derivative_weight = (2 / n) * np.sum(error * x)
            derivative_bias = (2 / n) * np.sum(error)

            self.weight = self.weight - self.learning_rate * derivative_weight
            self.bias = self.bias - self.learning_rate * derivative_bias

            sse = np.sum(error ** 2)
            self.sse_history.append(sse)

    def predict(self, x):
        x = np.array(x)
        return self.weight * x + self.bias

    def plot(self, x, y):
        x = np.array(x)
        y = np.array(y)

        plt.figure(figsize=(10, 8))

        plt.subplot(1, 2, 1)
        plt.plot(
            range(1, self.n_iters + 1),
            self.sse_history
        )
        plt.xlabel("Iteration")
        plt.ylabel("SSE")
        plt.title("SSE over Iterations")
        plt.grid()

        plt.subplot(1, 2, 2)
        plt.scatter(
            x,
            y,
            label="Data Points"
        )

        y_pred = self.predict(x)

        plt.plot(
            x,
            y_pred,
            label="Regression Line"
        )

        plt.xlabel("House Size (m²)")
        plt.ylabel("House Price (Thousands)")
        plt.title("Linear Regression")
        plt.legend()
        plt.grid()

        plt.tight_layout()
        plt.show()
```

```python
X = np.array([50, 60, 70, 80, 90])

y = np.array([150, 180, 210, 240, 270])
```

```python
print("X:", X)
print("y:", y)
```

```python
model = LinearRegression(
    learning_rate=0.001,
    n_iters=100
)

model.fit(X, y)
```

```python
print("Weight =", model.weight)
print("Bias =", model.bias)
```

```python
house_size = 70

predicted_price = model.predict(house_size)

print(
    f"Predicted price for {house_size} m² = "
    f"{predicted_price:.2f} thousand"
)
```

```python
model.plot(X, y)
```

```python
small_lr_model = LinearRegression(
    learning_rate=0.00001,
    n_iters=100
)

small_lr_model.fit(X, y)
```

```python
large_lr_model = LinearRegression(
    learning_rate=0.01,
    n_iters=100
)

large_lr_model.fit(X, y)
```

```python
large_lr_model = LinearRegression(
    learning_rate=0.01,
    n_iters=100
)

large_lr_model.fit(X, y)
```

```python
print("\nLearning Rate Comparison")

print("\nSmall Learning Rate:")
print("Learning Rate:", small_lr_model.learning_rate)
print("Final SSE:", small_lr_model.sse_history[-1])
print("Bias:", small_lr_model.bias)
print("Weight:", small_lr_model.weight)

print("\nNormal Learning Rate:")
print("Learning Rate:", model.learning_rate)
print("Final SSE:", model.sse_history[-1])
print("Bias:", model.bias)
print("Weight:", model.weight)

print("\nLarge Learning Rate:")
print("Learning Rate:", large_lr_model.learning_rate)
print("Final SSE:", large_lr_model.sse_history[-1])
print("Bias:", large_lr_model.bias)
print("Weight:", large_lr_model.weight)
```
