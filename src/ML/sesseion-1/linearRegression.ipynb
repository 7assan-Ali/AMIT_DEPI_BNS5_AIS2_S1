```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([2, 2.8, 3.6, 4.5])
```

```python
w = 0
b = 0
alpha = 0.01
```

```python
num_iteration = 1000

# Store SSE values
sse_value = []

# Gradient Descent
for i in range(num_iteration):

    # Prediction
    y_hat = w * x + b

    # Gradients
    D_W = (2 / 4) * np.sum((y_hat - y) * x)
    D_b = (2 / 4) * np.sum(y_hat - y)

    # Update parameters
    w = w - alpha * D_W
    b = b - alpha * D_b

    # Prediction after update
    y_hat = w * x + b

    # Calculate SSE
    sse = np.sum((y_hat - y) ** 2)
    sse_value.append(sse)

    # Print every 100 iterations
    if (i + 1) % 100 == 0:
        print(
            f"Iteration {i + 1}: "
            f"w = {w:.4f}, "
            f"b = {b:.4f}, "
            f"SSE = {sse:.4f}"
        )
```

```python
plt.figure(figsize=(12, 5))

# SSE vs Iterations
plt.subplot(1, 2, 1)
plt.plot(range(1, num_iteration + 1), sse_value, label="SSE")
plt.xlabel("Iteration")
plt.ylabel("SSE")
plt.title("SSE vs Iterations")


# Regression Line
plt.subplot(1, 2, 2)
plt.scatter(x, y, label="Data points")
plt.plot(x, y_hat, label="Regression line")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression")


plt.tight_layout()
plt.show()
```
