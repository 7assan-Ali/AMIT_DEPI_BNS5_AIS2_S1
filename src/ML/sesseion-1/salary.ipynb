```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv('D:\DEPI\AMIT_DEPI_BNS5_AIS2_S1\src\ML\sesseion-1\DataSet\Salary_Data.csv')
df.head()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['YearsExperience'],df['Salary'])
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Years of Experience vs Salary')
plt.show()
```

```python
x=df.iloc[:,:-1].values
y=df.iloc[:,1]
```

```python
from sklearn.model_selection import train_test_split
train_x,test_x,train_y,test_y=train_test_split(x,y,test_size=0.2,random_state=0,shuffle=True)
train_x.shape,train_y.shape,test_x.shape,test_y.shape
```

```python
linear_regressor = LinearRegression()
linear_regressor.fit(train_x, train_y)
linear_regressor.predict(train_x)
```

```python
plt.scatter(train_x, train_y, color='blue', label='Training Data')
plt.plot(train_x, linear_regressor.predict(train_x), color='red', label='Regression Line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Linear Regression: Years of Experience vs Salary')
plt.legend()
plt.show()
```

```python
plt.scatter(test_x, test_y, color='blue', label='Test Data')
plt.plot(test_x, linear_regressor.predict(test_x), color='red', label='Regression Line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Linear Regression: Years of Experience vs Salary')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error,r2_score

y_pred=linear_regressor.predict(test_x)
mse=mean_squared_error(test_y,y_pred)
r2=r2_score(test_y,y_pred)
root_mse=np.sqrt(mse)
print(f'Mean Squared Error: {mse}')
print(f'R-squared: {r2}')
print(f'Root Mean Squared Error: {root_mse}')
```
