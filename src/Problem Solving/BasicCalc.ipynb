```python
class Solution:
    def calculate(self, s: str) -> int:
        stack = []
        num = 0
        operation = '+'

        for i, char in enumerate(s):
            if char.isdigit():
                num = num * 10 + int(char)

            if char in '+-*/' or i == len(s) - 1:
                
                if operation == '+':
                    stack.append(num)

                elif operation == '-':
                    stack.append(-num)

                elif operation == '*':
                    stack[-1] = stack[-1] * num

                elif operation == '/':
                    stack[-1] = int(stack[-1] / num)

                operation = char
                num = 0

        return sum(stack)
```
