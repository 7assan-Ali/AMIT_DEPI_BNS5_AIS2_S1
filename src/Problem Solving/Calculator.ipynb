```python
def Calculator():
    """
    A simple command-line calculator that allows the user to perform
    basic arithmetic operations.

    The calculator supports the following operations:
    - Addition
    - Subtraction
    - Multiplication
    - Division

    The program runs continuously until the user selects the Exit option.

    :raises ValueError:
        If the user enters a non-numeric value when prompted for numbers.

    :raises ZeroDivisionError:
        Prevented by checking if the divisor is zero before performing division.

    :return:
        None.

    :rtype:
        None
    """

    while True:

        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Exit")

        choice = input("Choose: ")
        print("You chose:", choice)

        if choice == "5":
            print("Exiting the calculator.")
            break

        if choice not in ("1", "2", "3", "4"):
            print("Invalid Choice")
            continue

        num1 = float(input("First number: "))
        num2 = float(input("Second number: "))
        
        print("You entered:", num1, "and", num2)

        if choice == "1":
            result = num1 + num2
            print(f"Adding {num1} + {num2} = {result}")

        elif choice == "2":
            result = num1 - num2
            print(f"Subtracting {num1} - {num2} = {result}")

        elif choice == "3":
            result = num1 * num2
            print(f"Multiplying {num1} * {num2} = {result}")

        elif choice == "4":
            if num2 == 0:
                print("Division by zero is not allowed.")
                continue
            result = num1 / num2
            print(f"Dividing {num1} / {num2} = {result}")

        print("Result =", result)


Calculator()
```
