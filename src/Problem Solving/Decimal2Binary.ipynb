```python
def decimal2binary(n: int) -> str:
    """
    Convert a decimal number to binary.
    """
    if n ==0:
        return "0"
    binary = ""
    while n>0:
        reminder = n%2
        binary = str(reminder) + binary
        n = n//2
    return binary
```

```python
decimal2binary(11)
decimal2binary(13)
decimal2binary(5)
decimal2binary(20)
decimal2binary(2)

```
