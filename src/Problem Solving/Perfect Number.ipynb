```python
def Perfect_Number(n: int):
    """_summary_

    Args:
        n (int): _description_
    """ 
    sum =0
    for i in range(1,n):
        if n%i==0:
            sum+=i
            print(i)
    if sum==n:
        print(f"{n} is a perfect number")
    
```

```python
Perfect_Number(28)
```
