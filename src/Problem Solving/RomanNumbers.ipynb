```python
class Solution:
    def romanToInt(self, s: str) -> int:
        values = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }

        total_len=0

        for i in range(len(s)):
            # if i < len(n) - 1 and values[s[i]] < values[s[i + 1]]:
            if i < len(s) - 1 and values[s[i]] < values[s[i + 1]]:

                total_len-=values[s[i]]

            else:
                total_len+=values[s[i]]

        return total_len
        
```
