```python
def is_prime(n: int) -> bool:
    """
    Check if a number is prime.
    """
    if n<=1:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
```

```python
for i in range(1,1000):
    if is_prime(i) and is_prime(i+2):
        print(f"{i} and {i+2} are both prime numbers.")
    
```
