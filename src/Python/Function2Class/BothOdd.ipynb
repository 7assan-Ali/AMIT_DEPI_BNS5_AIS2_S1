```python
class BothOdd:
    """
    Find twin prime numbers.
    """

    def __init__(self, num):
        self.num = num

    def is_prime(self, num):
        """Check whether a number is prime."""

        if num <= 1:
            return False

        for i in range(2, num):
            if num % i == 0:
                return False

        return True

    def get_result(self):
        """Find twin prime numbers."""

        for num in range(1, self.num):
            if self.is_prime(num) and self.is_prime(num + 2):
                print(f"{num} and {num + 2} are both prime numbers.")
```

```python
B1 = BothOdd(20)
B1.get_result()
```

```python
import unittest


class TestBothOdd(unittest.TestCase):

    def setUp(self):
        self.twins = BothOdd(20)

    def test_prime_number(self):
        self.assertTrue(self.twins.is_prime(7))

    def test_not_prime_number(self):
        self.assertFalse(self.twins.is_prime(8))

    def test_prime_3(self):
        self.assertTrue(self.twins.is_prime(3))

    def test_one_is_not_prime(self):
        self.assertFalse(self.twins.is_prime(1))

    def test_zero_is_not_prime(self):
        self.assertFalse(self.twins.is_prime(0))


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
