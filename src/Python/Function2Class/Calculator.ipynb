```python
class Calculator:
    """
    Simple calculator for basic arithmetic operations.
    """

    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def add(self):
        return self.x + self.y

    def sub(self):
        return self.x - self.y

    def multi(self):
        return self.x * self.y

    def div(self):
        if self.y == 0:
            return ZeroDivisionError
        return self.x / self.y

    def operation_type(self):

        while True:
            print("\n1. Add")
            print("2. Subtract")
            print("3. Multiply")
            print("4. Divide")
            print("5. Exit")

            choice = input("Choose: ")

            if choice == "5":
                print("Goodbye!")
                break

            if choice not in ("1", "2", "3", "4"):
                print("Invalid choice!")
                continue

            self.x = float(input("First number: "))
            self.y = float(input("Second number: "))

            if choice == "1":
                print("Your Operation is Add:")
                print(self.x, "+", self.y)
                result = self.add()

            elif choice == "2":
                print("Your Operation is Subtract:")
                print(self.x, "-", self.y)
                result = self.sub()

            elif choice == "3":
                print("Your Operation is Multiply:")
                print(self.x, "*", self.y)
                result = self.multi()

            elif choice=="4":
                print("Your Operation is Divide:")
                print(self.x, "/", self.y)
                result = self.div()
            
            else:
                print("Invalid choice!")
            print("Result:", result)


```

```python
calc = Calculator()
calc.operation_type()
```

```python
import unittest


class TestCalculator(unittest.TestCase):

    def setUp(self):
        self.calc = Calculator()

    def test_add(self):
        self.calc.x = 10
        self.calc.y = 5
        self.assertEqual(self.calc.add(), 15)

    def test_sub(self):
        self.calc.x = 10
        self.calc.y = 5
        self.assertEqual(self.calc.sub(), 5)

    def test_multi(self):
        self.calc.x = 10
        self.calc.y = 5
        self.assertEqual(self.calc.multi(), 50)

    def test_div(self):
        self.calc.x = 10
        self.calc.y = 5
        self.assertEqual(self.calc.div(), 2)

    def test_divide_by_zero(self):
        self.calc.x = 10
        self.calc.y = 0
        self.assertEqual(self.calc.div(), ZeroDivisionError)


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
