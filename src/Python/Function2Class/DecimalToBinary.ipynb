```python
class DecimalTwoBinary:
    """
    Convert a decimal number to binary.
    """

    def __init__(self, decimal_num):
        self.decimal_num = decimal_num

    def decimal2binary(self):
        """Convert decimal number to binary."""

        if self.decimal_num == 0:
            return "0"

        binary = ""
        num = self.decimal_num

        while num > 0:
            remainder = num % 2
            binary = str(remainder) + binary
            num //= 2

        return binary
```

```python
d1=DecimalTwoBinary(12)
d2=DecimalTwoBinary(2)
d3=DecimalTwoBinary(7)
d4=DecimalTwoBinary(20)

print(d1.decimal2binary())
print(d2.decimal2binary())
print(d3.decimal2binary())
print(d4.decimal2binary())

```

```python
import unittest


class TestDecimalTwoBinary(unittest.TestCase):

    def test_binary_10(self):
        d1 = DecimalTwoBinary(10)
        self.assertEqual(d1.decimal2binary(), "1010")

    def test_binary_13(self):
        d1 = DecimalTwoBinary(13)
        self.assertEqual(d1.decimal2binary(), "1101")

    def test_binary_5(self):
        d1 = DecimalTwoBinary(5)
        self.assertEqual(d1.decimal2binary(), "101")

    def test_binary_20(self):
        d1 = DecimalTwoBinary(20)
        self.assertEqual(d1.decimal2binary(), "10100")

    def test_binary_2(self):
        d1 = DecimalTwoBinary(2)
        self.assertEqual(d1.decimal2binary(), "10")


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
