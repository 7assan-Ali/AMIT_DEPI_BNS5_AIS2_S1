```python
class PerfectNumber:

    def __init__(self, num):
        self.num = num

    def perfect_number(self):

        total = 0

        for i in range(1, self.num):
            if self.num % i == 0:
                total += i
                print(i)

        return total == self.num
```

```python
P1=PerfectNumber(12)
P2=PerfectNumber(28)

print(P1.perfect_number())
print(P2.perfect_number())
```

```python
import unittest


class TestPerfectNumber(unittest.TestCase):

    def test_perfect_6(self):
        p1 = PerfectNumber(6)
        self.assertTrue(p1.perfect_number())

    def test_perfect_28(self):
        p2 = PerfectNumber(28)
        self.assertTrue(p2.perfect_number())

    def test_not_perfect_12(self):
        p3 = PerfectNumber(12)
        self.assertFalse(p3.perfect_number())

    def test_not_perfect_10(self):
        p4 = PerfectNumber(10)
        self.assertFalse(p4.perfect_number())


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
