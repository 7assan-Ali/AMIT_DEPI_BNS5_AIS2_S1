```python
class PlayerName:

    def __init__(self, name):
        self.name = name

    def player_name(self):
        if self.name[0].lower() == "r":
            return self.name + " plays banjo"

        return self.name + " does not play banjo"
```

```python
P1=PlayerName("Ramy")
P2=PlayerName("Hassan")

print(P1.player_name())
print(P2.player_name())
```

```python
import unittest


class TestPlayerName(unittest.TestCase):

    def test_plays_banjo(self):
        p1 = PlayerName("Rick")
        self.assertEqual(
            p1.player_name(),
            "Rick plays banjo"
        )

    def test_does_not_play_banjo(self):
        p2 = PlayerName("John")
        self.assertEqual(
            p2.player_name(),
            "John does not play banjo"
        )

    def test_lowercase_r(self):
        p3 = PlayerName("robert")
        self.assertEqual(
            p3.player_name(),
            "robert plays banjo"
        )


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
