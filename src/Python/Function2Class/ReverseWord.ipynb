```python
class ReverseWords:

    def __init__(self, message):
        self.message = message

    def reverse_words(self):
        words = self.message.split()
        reversed_words = words[::-1]

        return " ".join(reversed_words)
```

```python
message = "The greatest victory is that which requires no battle"

m1 = ReverseWords(message)

print(m1.reverse_words())
```

```python
import unittest


class TestReverseWords(unittest.TestCase):

    def setUp(self):
        self.message = "The greatest victory is that which requires no battle"
        self.m1 = ReverseWords(self.message)

    def test_reverse_words(self):
        self.assertEqual(
            self.m1.reverse_words(),
            "battle no requires which that is victory greatest The"
        )


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
