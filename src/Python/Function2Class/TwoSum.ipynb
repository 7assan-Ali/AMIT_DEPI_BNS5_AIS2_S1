```python
class TwoSum:

    def __init__(self, nums, target):
        self.nums = nums
        self.target = target

    def get_result(self):

        for i in range(len(self.nums)):
            for j in range(i + 1, len(self.nums)):

                if self.nums[i] + self.nums[j] == self.target:
                    return [i, j]
```

```python
T1 = TwoSum([2, 7, 11, 15], 9)

print(T1.get_result())
```
