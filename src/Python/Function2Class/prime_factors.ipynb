```python
class PrimeFactors:

    def __init__(self, num):
        self.num = num

    def is_prime(self, num):
        """
        Check whether a number is prime.

        Returns:
            bool: True if the number is prime, otherwise False.
        """

        if num <= 1:
            return False

        for i in range(2, num):
            if num % i == 0:
                return False

        return True

    def prime_factors(self):
        """
        Find the prime factors of the number.

        Returns:
            list: A list containing the prime factors.
        """

        prime_factors_list = []
        n = self.num

        for i in range(2, n + 1):
            if self.is_prime(i):
                while n % i == 0:
                    prime_factors_list.append(i)
                    n //= i

        return prime_factors_list
```

```python
import unittest


class TestPrimeFactors(unittest.TestCase):

    def setUp(self):
        self.number = PrimeFactors(60)

    def test_is_prime_true(self):
        self.assertTrue(self.number.is_prime(7))

    def test_is_prime_false(self):
        self.assertFalse(self.number.is_prime(10))

    def test_prime_factors(self):
        self.assertEqual(
            self.number.prime_factors(),
            [2, 2, 3, 5]
        )


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
