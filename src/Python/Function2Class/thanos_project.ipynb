```python
import os
import random


class ThanosProject:

    def __init__(self, folder):
        self.folder = folder

    def thanos_project(self):

        # Create 10 files
        for i in range(1, 11):
            file_path = os.path.join(
                self.folder,
                f"file{i}.txt"
            )

            with open(file_path, "w") as f:
                f.write(f"This is file {i}")

        # Get files
        files = os.listdir(self.folder)

        print("Files before delete:", len(files))

        # Delete half randomly
        delete_files = random.sample(
            files,
            len(files) // 2
        )

        for file in delete_files:
            os.remove(
                os.path.join(self.folder, file)
            )

        # Remaining files
        files = os.listdir(self.folder)

        print("Files after delete:", len(files))

        return files
```

```python
folder = r"D:\DEPI\AMIT_DEPI_BNS5_AIS2_S1\src\Python\Function2Class"

thanos = ThanosProject(folder)

remaining_files = thanos.thanos_project()

print("Remaining files:", remaining_files)
```

```python
import unittest
import tempfile
import os


class TestThanosProject(unittest.TestCase):

    def test_thanos_project(self):

        with tempfile.TemporaryDirectory() as folder:

            thanos = ThanosProject(folder)

            remaining_files = thanos.thanos_project()

            # Should remain 5 files
            self.assertEqual(len(remaining_files), 5)

            # All remaining files should exist
            for file in remaining_files:
                self.assertTrue(
                    os.path.exists(
                        os.path.join(folder, file)
                    )
                )


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
