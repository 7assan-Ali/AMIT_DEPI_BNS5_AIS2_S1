```python
from abc import ABC ,abstractmethod


class Animals(ABC):
    @abstractmethod
    
    def make_sound(self):
        pass
    
    def animal_descripe(self):
        print("This is animal.")
        
class Dog(Animals):
    def make_sound(self):
        return "Woof"
    
    
    
class Cat(Animals):
    def make_sound(self):
        return "Meow"
    
    
    
class Cow(Animals):
    def make_sound(self):
        return "Moo"
    
   
    
    
    
```

```python
dog = Dog()
cat = Cat()
cow = Cow()

print("Dog:")
print(dog.make_sound())
dog.animal_descripe()

print("\nCat:")
print(cat.make_sound())
cat.animal_descripe()

print("\nCow:")
print(cow.make_sound())
cow.animal_descripe()

```
