```python
class Calculator:

    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def add(self):
        return self.x + self.y

    def sub(self):
        return self.x - self.y

    def multi(self):
        return self.x * self.y

    def div(self):
        if self.y == 0:
            return "Error: Cannot divide by zero."

        return self.x / self.y

    def calculator(self):

        while True:

            print("\n========== Calculator ==========")
            print("1: Add")
            print("2: Subtract")
            print("3: Multiply")
            print("4: Divide")
            print("5: Exit")

            choice = input(
                "Enter the number of the operation you want to perform (1/2/3/4/5): "
            )

            # Exit
            if choice == "5":
                print("GoodBye!")
                break

            # Validate choice
            if choice not in ["1", "2", "3", "4"]:
                print("The choice is not valid. Enter a choice from 1 to 5.")
                continue

            # Get numbers
            try:
                self.x = float(input("Enter first value: "))
                self.y = float(input("Enter second value: "))

            except ValueError:
                print("Please enter numbers only.")
                continue

            # Operations
            if choice == "1":
                print("\nYour Operation is Add:")
                print(self.x, "+", self.y)
                result = self.add()

            elif choice == "2":
                print("\nYour Operation is Subtract:")
                print(self.x, "-", self.y)
                result = self.sub()

            elif choice == "3":
                print("\nYour Operation is Multiply:")
                print(self.x, "*", self.y)
                result = self.multi()

            elif choice == "4":
                print("\nYour Operation is Divide:")
                print(self.x, "/", self.y)
                result = self.div()

            print(f"Result = {result}")
```

```python
calc = Calculator()

calc.calculator()
```
