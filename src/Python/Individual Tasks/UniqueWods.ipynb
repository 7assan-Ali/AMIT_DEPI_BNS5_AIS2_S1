```python
def unique_words(words):
    unique_words=set(words)
    counts={}
    
    for i in unique_words:
        for j in words:
            if i==j:
                if i in counts:
                    counts[i]+=1
                else:
                    counts[i]=1
    return counts
```

```python
words = [
    "Welcome",
    "Ali",
    "Hi",
    "Ali",
    "No",
    "Hi",
    "No",
    "Ali",
    "No",
    "Ali"
]

result = unique_words(words)

print(result)
```
