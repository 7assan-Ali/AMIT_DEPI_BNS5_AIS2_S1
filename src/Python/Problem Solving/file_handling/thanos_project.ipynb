```python
import os
import random

def thanos_project():

    folder = r"D:\DEPI\AMIT_DEPI_BNS5_AIS2_S1\src\Python\Problem Solving\file_handling"

    for i in range(1, 11):
        with open(os.path.join(folder, f"file{i}.txt"), "w") as f:
            f.write(f"This is file {i}")

    files = os.listdir(folder)

    print("Files before delete:", len(files))

    delete_files = random.sample(files, len(files) // 2)

    for file in delete_files:
        os.remove(os.path.join(folder, file))

    files = os.listdir(folder)

    print("Files after delete:", len(files))
    print("Remaining files:")

    for file in files:
        print(file)

thanos_project()
```
