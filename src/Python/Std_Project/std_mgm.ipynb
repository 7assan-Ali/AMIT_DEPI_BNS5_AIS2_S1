```python
class Student:
    _id_counter=1
    def __init__(self,name):
         self.student_Id=Student._id_counter
         Student._id_counter += 1
         self.name=name  
         self.grades={}
         self.enrolled_courses=[]
         
    def __str__(self):
     return f"Student ID :{self.student_Id} \nName : {self.name}\nGrades:{self.grades}\nCourses:{self.enrolled_courses}"
 
    def __repr__(self)->str:
     return f"Student ID :{self.student_Id} \nName : {self.name}\nGrades:{self.grades}\nCourses:{self.enrolled_courses}"
 
    def add_grade(self, course_id, grades):
        if not 0<=grades<=100:
            raise ValueError("Grade must between 0 and 100")
        self.grades[course_id]=grades
        
    def enroll_in_course(self,course):
        if course in self.enrolled_courses:
            raise ValueError ("Student already enrolled in this course")
        else:
            self.enrolled_courses.append(course)   
    
```

```python
class Course:
    _id_counter = 1

    def __init__(self, name):
        self.course_id = Course._id_counter
        Course._id_counter += 1
        self.name = name
        self.enrolled_students = []

    def __str__(self):
        return f"Course ID: {self.course_id}, Name: {self.name}, Enrolled: {len(self.enrolled_students)}"

    # تسجيل الطالب في الكورس
    def enroll_student(self, student):
        if student not in self.enrolled_students:
            self.enrolled_students.append(student)
            print(f"Student {student.name} enrolled successfully in {self.name}.")
        else:
            print(f"Student {student.name} is already enrolled in {self.name}.")

    # إزالة الطالب من الكورس
    def remove_student(self, student):
        if student in self.enrolled_students:
            self.enrolled_students.remove(student)
            print(f"Student {student.name} removed from {self.name}.")
        else:
            print(f"Student {student.name} is not enrolled in {self.name}.")
```

```python
import unittest

class TestCourse(unittest.TestCase):

    def setUp(self):
        self.course = Course("Machine Learning")
        self.student = Student("Hassan")

    def test_creation(self):
        self.assertEqual(self.course.name, "Machine Learning")
        self.assertEqual(len(self.course.enrolled_students), 0)

    def test_enroll_student(self):
        self.course.enroll_student(self.student)

        self.assertIn(
            self.student,
            self.course.enrolled_students
        )

    def test_student_enrolled_once(self):
        self.course.enroll_student(self.student)
        self.course.enroll_student(self.student)

        self.assertEqual(
            len(self.course.enrolled_students),
            1
        )

    def test_remove_student(self):
        self.course.enroll_student(self.student)
        self.course.remove_student(self.student)

        self.assertNotIn(
            self.student,
            self.course.enrolled_students
        )

    def test_remove_student_not_enrolled(self):
        self.course.remove_student(self.student)

        self.assertNotIn(
            self.student,
            self.course.enrolled_students
        )

    def test_course_string(self):
        result = str(self.course)

        self.assertIn("Machine Learning", result)
        self.assertIn("Enrolled: 0", result)


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
