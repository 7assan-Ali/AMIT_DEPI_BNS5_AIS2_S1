```python
Encoded_message="&&&**$gnirtS PLIO!!@1234"
Encoded_message
```

```python
start_postion=Encoded_message.index("g")
start_postion
```

```python
length_of_message=len(Encoded_message)
length_of_message
```

```python
end_postion=length_of_message-start_postion
end_postion
```

```python
core=Encoded_message[start_postion:end_postion-1]
core
```

```python
space=core.find(" ")
space
```

```python
first_word = core[:space]
first_word
```

```python
reserve_first_word=first_word[::-1]
reserve_first_word
```

```python
second_word = core[space+1:]
second_word
```

```python
second_word = second_word.replace("I", "E")
second_word = second_word.replace("O", "U")
second_word
```
