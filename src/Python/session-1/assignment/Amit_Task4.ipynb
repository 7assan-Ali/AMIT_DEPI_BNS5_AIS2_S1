```python
Encoded_message = "##$$$@!yalpstcejorp EPUVT****9887"
Encoded_message
```

```python
start_postion=Encoded_message.index("y")
start_postion
```

```python
length_of_message=len(Encoded_message)
length_of_message
```

```python
end_postion=length_of_message-start_postion
end_postion
```

```python
core=Encoded_message[start_postion:end_postion-1]
core
```

```python
space=core.find(" ")
space
```

```python
first_word = core[:space]
first_word
```

```python
reverse_first_word=first_word[::-1]
reverse_first_word
```

```python
second_word = core[space+1:]
second_word
```

```python
second_word = second_word.replace("E", "A")
second_word = second_word.replace("U", "O")
second_word
```

```python
final_decoded=reverse_first_word+" "+second_word
final_decoded
```
