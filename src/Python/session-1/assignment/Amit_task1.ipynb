```python
User_email=input(str("Enter your email address: "))
```

```python
count_of_at=User_email.count("@")
count_of_at
```

```python
count_of_dot=User_email.count(".")
count_of_dot
```

```python
at_position = User_email.index("@")
at_position
```

```python
dot_position = User_email.rfind(".")
dot_position
```

```python
if count_of_at == 1 and dot_position > at_position:
    print("Valid Email")
    username = User_email[:at_position]
    print("Username = ",username)
    domain=User_email[at_position+1:dot_position]
    print("Domain = ",domain)
else:
    print("Invalid Email")
```

```python
domain_type=User_email[-4:]
domain_type
```

```python
if domain_type==".com":
    print("Commercial Domain")
elif domain_type==".edu":
    print("Educational Domain")
else:
    print("Not Defined")
```
