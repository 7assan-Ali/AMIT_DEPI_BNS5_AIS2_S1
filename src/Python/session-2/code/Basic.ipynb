```python
is_sunny=True
temperature=30
if is_sunny:
    print("It is a sunny day")
    if temperature > 25:
        print("Temperature is good for swimming")
    else:
        print("Temperature is not good for swimming")
else:
    print("It is not a sunny day")
```
