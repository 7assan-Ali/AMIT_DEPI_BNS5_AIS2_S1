```python
def add(x: float , y:float):
    '''
    sum funtion 
    
    args:
        param_1= user must input the first num
        type_param_1: float
        param_2= user must input the sconed num
        type_param_2: float
         retutn : this function return sum of two nums
         return type: float        
    
    '''
    return x+y
```

```python
def sub(x: float , y:float):
    '''
    sub funtion 
    
    args:
        param_1= user must input the first num
        type_param_1: float
        param_2= user must input the sconed num
        type_param_2: float
         retutn : this function return sub of two nums
         return type: float        
    
    '''
    return x-y
```

```python
def mult(x: float , y:float):
    '''
    multiply funtion 
    
    args:
        param_1= user must input the first num
        type_param_1: float
        param_2= user must input the sconed num
        type_param_2: float
         retutn : this function return multiply of two nums
         return type: float        
    
    '''
    return x*y
```

```python
def div(x: float , y:float):
    '''
    divesion funtion 
    
    args:
        param_1= user must input the first num
        type_param_1: float
        param_2= user must input the sconed num
        type_param_2: float
         retutn : this function return div of two nums
         return type: float        
    
    '''
    return x/y
```

```python
print("===== Calculator App =====")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")

choice = input("Choose (1/2/3/4): ")

num_1 = float(input("Enter first number: "))
num_2 = float(input("Enter second number: "))

if choice == "1":
    print(f"first num is :{num_1}, second num is :{num_2}  result is", add(num_1, num_2))

elif choice == "2":
    print("first num is :{num_1}, second num is :{num_2}  result is", sub(num_1, num_2))

elif choice == "3":
    print("first num is :{num_1}, second num is :{num_2}  result is", mult(num_1, num_2))

elif choice == "4":
    print("first num is :{num_1}, second num is :{num_2}  result is", div(num_1, num_2))

else:
    print("Invalid choice!")
```
