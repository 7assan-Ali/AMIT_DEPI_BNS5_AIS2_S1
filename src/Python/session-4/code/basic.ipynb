```python
def common_divisor(num1:int,num2:int) -> int:
   limit = min(num1,num2)
   divisors = []
   
   for divisor in range(1,limit+1):
       if num1%divisor==0 and num2%divisor==0:
           divisors.append(divisor)
   return divisors
```

```python
def reversed_string(s:str) -> list:
    """
    Reverse a string.
    
    Parameters
    s : str
    String to reverse.
    
    Returns
    str
    Reversed string.
    """
    words = s.split()
    words.reverse()
    return words
```

```python
import string
import random
import secrets

def gen_password(length: int=12) -> str:
    """ 
    Generate a random password of specified length.
    
    Parameters
    ----------
    length : int, optional
        Length of the password to generate (default is 12).
    
    Returns
    -------
    str
        Randomly generated password.
    """
    if length <= 0:
       raise ValueError("Password length must be a positive integer.")
    char=(
    string.ascii_letters + string.digits + string.punctuation
    )
    
    return "".join(secrets.choice(char) for _ in range(length))
   
   
   
```

```python
words=["welcome","Ali","hi","Ali","Hello","Ali","hi","Ali","Hello","Ali"
       ,"hi","Ali","Hello","Ali","hi","Ali","Hello","Ali","hi","Ali"]


def words_count(words: list) -> dict:
    """
    Count the occurrences of each unique word in a list.
    
    Parameters
    ----------
    words : list
        List of words to count.
    
    Returns
    -------
    dict
        Dictionary with unique words as keys and their counts as values.
    """
    unique_words = set(words)
    count_dict = {word: words.count(word) for word in unique_words}
    for i in unique_words:
        for j in words:
            if i==j:
                count_dict[i]+=1
    return count_dict
```

## Chatbot


```python
import random

response={
    
    "hello": ["Hello!", "Hi there! How can I help you today?", "Greetings! What can I do for you?"],
    "how are you?": ["I'm just a bot, but I'm doing great! How about you?", "I'm doing well, thank you for asking! How about you?"],
    "what's your name?": ["I'm a simple chatbot created to assist you.", "You can call me Chatbot!"],
    "goodbye": ["Goodbye! Have a great day!", "See you later!"],
    "default": ["I'm sorry, I don't understand. Can you please rephrase your question?", "Could you clarify that please?"]
}

def get_response(user_input: str) -> str:
    """
    Get a response from the chatbot based on user input.
    
    Parameters
    ----------
    user_input : str
        The input string from the user.
    
    Returns
    -------
    str
        A response string from the chatbot.
    """
    user_input = user_input.lower()
    if user_input in response:
        return random.choice(response[user_input])
    else:
        return random.choice(response["default"])
    
def chatbot():
    """
    Run the chatbot in a loop until the user types 'exit'.
    """
    print("Welcome to the Chatbot! Type 'exit' to end the conversation.")
    while True:
        user_input = input("You: ")
        if user_input.lower() == "exit":
            print("Chatbot: Goodbye!")
            break
        response = get_response(user_input)
        print(f"Chatbot: {response}")


```
